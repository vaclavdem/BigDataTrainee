**8. Среди кого больше доля зарабатывающих много (>50K): среди женатых или холостых мужчин (признак *marital-status*)? Женатыми считаем тех, у кого *marital-status* начинается с *Married* (Married-civ-spouse, Married-spouse-absent или Married-AF-spouse), остальных считаем холостыми.**

```python
married_status = ['Married-civ-spouse', 'Married-spouse-absent', 'Married-AF-spouse']
non_married_status = ['Divorced', 'Never-married', 'Separated', 'Widowed']

married = data[data['marital-status'].isin(married_status)]
non_married = data[data['marital-status'].isin(non_married_status)]
man_married = married[married['sex'] == 'Male']
man_non_married = non_married[non_married['sex'] == 'Male']
percent_more_salary_married = len(man_married[man_married['salary'] == '>50K']) / len(man_married) * 100
percent_more_salary_non_married = len(man_non_married[man_non_married['salary'] == '>50K']) / len(man_non_married) * 100

print(percent_more_salary_married, percent_more_salary_non_married)
```

**12-13. Определите количество зарабатывающих >50K в каждой из возрастных групп (колонка AgeGroup), а также выведите название возрастной группы, в которой чаще зарабатывают больше 50К (>50K)**

**14. Сгруппируйте людей по типу занятости (колонка occupation) и определите количество людей в каждой группе. После чего напишите функциюю фильтрации filter_func, которая будет возвращать только те группы, в которых средний возраст (колонка age) не больше 40 и в которых все работники отрабатывают более 5 часов в неделю (колонка hours-per-week)**

```python
grouped_data = data.groupby('occupation')
grouped_data_size = grouped_data.size()
print(grouped_data_size)

def filter_func(grouped_data):
    filtered_groups = []
    for group_name, group_df in grouped_data:
        if group_df['age'].mean() <= 40 and group_df['hours-per-week'].min() > 5:
            filtered_groups.append(group_df)
    filtered_groups = pd.concat(filtered_groups)
    filtered_groups = filtered_groups.groupby('occupation').size()
    return filtered_groups

filtered_grouped_data = filter_func(grouped_data)
print(filtered_grouped_data)
```
