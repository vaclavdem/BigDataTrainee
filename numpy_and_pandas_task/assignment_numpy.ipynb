**1.** Дан случайный массив, поменять знак у элементов, значения которых между 3 и 8

```python
arr = np.random.randint(-3, 14, size=10)
print("Исходный массив:", arr)

for i in range(len(arr)):
    if 3 <= arr[i] and arr[i] <= 8:
        arr[i] = -arr[i]

print("Результат:", arr)
```

**2.** Заменить максимальный элемент случайного массива на 0

```python
arr = np.random.randint(-3, 14, size=10)
print("Исходный массив:", arr)

max = np.max(arr)

for i in range(len(arr)):
    if arr[i] == max:
        arr[i] = 0

print("Результат:", arr)
```

**3.** Построить прямое произведение массивов (все комбинации с каждым элементом). На вход подается двумерный массив

```python
arr_1 = np.random.randint(-3, 14, size=(2, 3))
print("Исходный массив номер 1:\n", arr_1)

arr_2 = np.empty((9, 2), dtype=int)

len_1 = len(arr_1[0])
len_2 = len(arr_1[1])

for i in range(len_1):
    for j in range(len_2):
        arr_2[i * len_1 + j][0] = arr_1[0][i]
        arr_2[i * len_1 + j][1] = arr_1[1][j]

print("Результат:\n", arr_2)
```

**4.** Даны 2 массива A (8x3) и B (2x2). Найти строки в A, которые содержат элементы из каждой строки в B, независимо от порядка элементов в B

```python
arr_1 = np.random.randint(-3, 14, size=(8, 3))
print("Исходный массив номер 1:\n", arr_1)

arr_2 = np.random.randint(-3, 14, size=(2, 2))
print("Исходный массив номер 2:\n", arr_2)

print("Результат:")
for i in range(len(arr_1)):
    ok = True
    for j in range(len(arr_2)):
        mask = np.isin(arr_1[i], arr_2[j])
        if True not in mask:
            ok = False
    if ok:
        print(arr_1[i])
```

**5.** Дана 10x3 матрица, найти строки из неравных значений (например строка [2,2,3] остается, строка [3,3,3] удаляется)

```python
arr_1 = np.random.randint(1, 5, size=(10, 3))
print("Исходный массив номер 1:\n", arr_1)

print("Результат:\n")
for i in range(len(arr_1)):
    first_al = arr_1[i][0]
    ok = True
    for j in range(len(arr_1[i])):
        if arr_1[i][j] != first_al:
            ok = False
    if ok:
        print(arr_1[i])
```

* __Задача 1__: Подсчитать произведение ненулевых элементов на диагонали прямоугольной матрицы.  
 Например, для X = np.array([[1, 0, 1], [2, 0, 2], [3, 0, 3], [4, 4, 4]]) ответ 3.

```python
x_com = [[1, 0, 1], [2, 0, 2], [3, 0, 3], [4, 4, 4]]
x_np = np.array([[1, 0, 1], [2, 0, 2], [3, 0, 3], [4, 4, 4]])

#common realization
result = 1
for i in range(min(len(x_com), len(x_com[0]))):
    if x_com[i][i] != 0:
        result *= x_com[i][i]
print("common realization: ", result)

#np realization
diag = np.diagonal(x_np)
nonzero_diag = diag[diag != 0]
prod = np.prod(nonzero_diag)
print("np realization: ", prod)
```

* __Задача 2__: Даны два вектора x и y. Проверить, задают ли они одно и то же мультимножество.  
  Например, для x = np.array([1, 2, 2, 4]), y = np.array([4, 2, 1, 2]) ответ True.

```python
x = [1, 2, 2, 4]
y = [4, 2, 1, 2]
x_np = np.array([1, 2, 2, 4])
y_np = np.array([4, 2, 1, 2])

#common realization
same_multiset = x.sort() == y.sort()
print("common realization: ", same_multiset)

#np realization
same_multiset = np.array_equal(np.sort(x_np), np.sort(y_np))
print("np realization: ", same_multiset)
```

* __Задача 3__: Найти максимальный элемент в векторе x среди элементов, перед которыми стоит ноль. 
 Например, для x = np.array([6, 2, 0, 3, 0, 0, 5, 7, 0]) ответ 5.

```python
x = [6, 2, 0, 3, 0, 0, 5, 7, 0]
x_np = np.array([6, 2, 0, 3, 0, 0, 5, 7, 0])

#common realization
max = -1
for i in range(len(x)):
    if i > 0 and x[i - 1] == 0 and x[i] > max:
        max = x[i]
print("common realization: ", max)

#np realization
mask = x_np[:-1] == 0
candidates = x_np[1:][mask]
result = np.max(candidates)
print("np realization: ", result)
```

* __Задача 4__: Реализовать кодирование длин серий (Run-length encoding). Для некоторого вектора x необходимо вернуть кортеж из двух векторов одинаковой длины. Первый содержит числа, а второй - сколько раз их нужно повторить.  
 Например, для x = np.array([2, 2, 2, 3, 3, 3, 5]) ответ (np.array([2, 3, 5]), np.array([3, 3, 1])).

```python
x = [2, 2, 2, 3, 3, 3, 5]
x_np = np.array([2, 2, 2, 3, 3, 3, 5])

#common realization
y_val = []
y_count = []
previous = x[0]
count = 1
for i in range(1, len(x)):
    if x[i] != previous or i == len(x) - 1:
        y_val.append(previous)
        y_count.append(count)
        previous = x[i]
        count = 1
    else:
        count += 1
y_val.append(previous)
y_count.append(count)
print("common realization: ", y_val, y_count)

#np realization
changes = np.where(np.diff(x_np) != 0)[0] + 1
counts = np.diff(np.concatenate(([0], changes, [x_np.size])))
values = x_np[np.concatenate(([0], changes))]
print("np realization: ", values, counts)
```

* __Задача 5__: Даны две выборки объектов - X и Y. Вычислить матрицу евклидовых расстояний между объектами. Сравните с функцией scipy.spatial.distance.cdist по скорости работы.

```python
from scipy.spatial.distance import cdist
import time

X = np.random.rand(2000, 50)
Y = np.random.rand(2000, 50)

#scipy realization

t0 = time.time()
D_scipy = cdist(X, Y, metric='euclidean')
t_scipy = time.time() - t0

print("scipy realization: \n", D_scipy)
print(t_scipy)

#np realization
def euclidean_distance_matrix(X, Y):
    X_sq = np.sum(X**2, axis=1, keepdims=True)   # (n, 1)
    Y_sq = np.sum(Y**2, axis=1, keepdims=True).T # (1, m)
    distances_sq = X_sq + Y_sq - 2 * X @ Y.T
    return np.sqrt(np.maximum(distances_sq, 0.0))

t0 = time.time()
D_numpy = euclidean_distance_matrix(X, Y)
t_numpy = time.time() - t0
    
print("np realization: \n", D_numpy)
print(t_numpy)
```
